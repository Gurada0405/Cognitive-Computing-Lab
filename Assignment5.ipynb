```python
x = np.array([1,2,3,4,5,1,2,1,1,1])
```

```python
values, counts = np.unique(x, return_counts=True)

most_frequent = values[np.argmax(counts)]

indices = np.where(x == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

```python
import numpy as np

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)

max_count = np.max(counts)

most_frequent = values[counts == max_count]

print("Most frequent value(s):", most_frequent)

for value in most_frequent:
    indices = np.where(y == value)[0]
    print("Value:", value, "Indices:", indices)
```

```python
import numpy as np

Swathi = np.linspace(10, 100, 25)

print("Array:", Swathi)

print("Dimensions:", Swathi.ndim)
print("Shape:", Swathi.shape)
print("Total elements:", Swathi.size)
print("Data type:", Swathi.dtype)
print("Total bytes:", Swathi.nbytes)

transpose = Swathi.reshape(1, 25)

print("Transpose using reshape():")
print(transpose)

print("Transpose using T:")
print(Swathi.T)
```

```python
import numpy as np

ucs420_Swathi = np.array([[10, 20, 30, 40],
                          [50, 60, 70, 80],
                          [90, 15, 20, 35]])

print("Original array:")
print(ucs420_Swathi)

print("Mean:", np.mean(ucs420_Swathi))

print("Median:", np.median(ucs420_Swathi))

print("Maximum:", np.max(ucs420_Swathi))

print("Minimum:", np.min(ucs420_Swathi))

print("Unique elements:", np.unique(ucs420_Swathi))

reshaped_ucs420_Swathi = ucs420_Swathi.reshape(4, 3)

print("Reshaped array (4 x 3):")
print(reshaped_ucs420_Swathi)

resized_ucs420_Swathi = np.resize(ucs420_Swathi, (2, 3))

print("Resized array (2 x 3):")
print(resized_ucs420_Swathi)
```
