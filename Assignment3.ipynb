```python
import pandas as pd

iris = pd.read_csv("Iris.csv")

print(iris.head())
```

```python
employees = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [4, 8, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01',
                     '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees_df = pd.DataFrame(employees)

employees_df.to_csv("employees.csv", index=False)

print("Employee Dataset:")
print(employees_df)
```

```python
print("\nQ6(a). Shape:")
print(employees_df.shape)
```

```python
print("\nQ6(b). DataFrame Information:")
print(employees_df.info())
```

```python
print("\nQ6(c). Descriptive Statistics:")
print(employees_df.describe())
```

```python
print("\nQ6(d). First 5 rows:")
print(employees_df.head(5))

print("\nLast 3 rows:")
print(employees_df.tail(3))
```

```python
average_salary = employees_df['Salary'].mean()
print("\nQ6(e)(i). Average Salary:")
print(average_salary)
```

```python
total_bonus = employees_df['Bonus'].sum()

print("\nQ6(e)(ii). Total Bonus:")
print(total_bonus)
```

```python
youngest_age = employees_df['Age'].min()

print("\nQ6(e)(iii). Youngest Employee Age:")
print(youngest_age)
```

```python
highest_rating = employees_df['Rating'].max()

print("\nQ6(e)(iv). Highest Performance Rating:")
print(highest_rating)
```

```python
sorted_df = employees_df.sort_values(
    by='Salary',
    ascending=False
)

print("\nQ6(f). Sorted by Salary (Descending):")
print(sorted_df)
```

```python
def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees_df['Performance'] = employees_df['Rating'].apply(
    performance_category
)

print("\nQ6(g). Performance Category:")
print(employees_df)
```

```python
print("\nQ6(h). Missing Values:")
print(employees_df.isnull().sum())
```

```python
employees_df = employees_df.rename(
    columns={'Employee_ID': 'ID'}
)
```

```python
print("\nQ6(i). After Renaming Employee_ID to ID:")
print(employees_df)
```

```python
more_than_5_years = employees_df[
    employees_df['Years_of_Experience'] > 5
]

print("\nQ6(j)(i). Employees with more than 5 years experience:")
print(more_than_5_years)
```

```python
it_employees = employees_df[
    employees_df['Department'] == 'IT'
]

print("\nQ6(j)(ii). Employees in IT Department:")
print(it_employees)
```

```python
employees_df['Tax'] = employees_df['Salary'] * 0.10

print("\nQ6(k). DataFrame with Tax:")
print(employees_df)
```

```python
employees_df.to_csv(
    "employees_modified.csv",
    index=False
)

print("\nQ6(l). Modified DataFrame saved as employees_modified.csv")
```
