```python
L = [10, 0, 20, 30, 40, 50, 60, 70, 80, 90]
scores = tuple(L[:8])
print("Score Tuples:", scores)
```

```python
try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)
    print("Tuples are immutable, unlike lists, so their elements cannot be changed.")
```

```python
first, second, *remaining = scores

print("First:", first)
print("Second:", second)
print("Remaining:", remaining)
```

```python
import random
random.seed(1024170106)
```

```python
numbers = [random.randint(100, 900) for _ in range(100)]
```

```python
print("numbers : ", numbers)
```

```python
odd_numbers = [x for x in numbers if x % 2 != 0]

print("Number of odd numbers:", len(odd_numbers))
print("Odd numbers:", odd_numbers)
```

```python
even_numbers = [x for x in numbers if x % 2 == 0]

print("Number of even numbers:", len(even_numbers))
print("Even numbers:", even_numbers)
```

```python
def is_prime(n):
    if n < 2:
        return False
    
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    
    return True
```

```python
prime_numbers = [x for x in numbers if is_prime(x)]

print("Number of prime numbers:", len(prime_numbers))
print("Prime numbers:", prime_numbers)
```

```python
most_frequent = max(set(numbers), key=numbers.count)

print("Most frequent number:", most_frequent)
print("Frequency:", numbers.count(most_frequent))
```

```python
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
```

```python
print("Symmetric Difference:", A.symmetric_difference(B))
```

```python
print("A is subset of B:", A.issubset(B))
print("B is superset of A:", B.issuperset(A))
```

```python
my_dict = {
    "name": "Sam",
    "roll_no": "12345678",
    "branch": "CSE",
    "age": 22,
    "city": "Patiala"
}

print("Original dictionary:")
print(my_dict)
```

```python
my_dict["location"] = my_dict.pop("city")
print(my_dict)
```

```python
my_dict["cgpa"] = 8.5

print(my_dict)
```

```python
my_dict["age"] += 1

print(my_dict)
```

```python
dict1 = my_dict.copy()
removed = dict1.pop("branch")

print("Using pop():", dict1)
print("Removed value:", removed)
```

```python
dict2 = my_dict.copy()
del dict2["branch"]

print("Using del:", dict2)
```

```python
for key, value in my_dict.items():
    print(key, "→", value)
```

```python
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key does not exist.")
```
