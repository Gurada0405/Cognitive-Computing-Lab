```python
paragraph = """Artificial intelligence is transforming the world through smart technologies.
Machine learning helps computers learn from data and improve their performance.
Natural language processing enables machines to understand human language.
AI applications are widely used in healthcare, education, and transportation.
Researchers are developing intelligent systems to solve complex problems.
Responsible AI can make technology more useful, accessible, and efficient."""

print(paragraph)
```

```python
import re

text_lower = paragraph.lower()
clean_text = re.sub(r'[^\w\s]', '', text_lower)

print("Lowercase text:\n", text_lower)
print("\nText after removing punctuation:\n", clean_text)
```

```python
import re

# Tokenize into words
word_tokens = clean_text.split()

# Tokenize into sentences using regular expressions
sentence_tokens = re.split(r'(?<=[.!?])\s+', paragraph.strip())

print("Word tokens:")
print(word_tokens)

print("\nSentence tokens:")
print(sentence_tokens)
```

```python
from nltk.tokenize import word_tokenize
import re

# Python split()
split_tokens = clean_text.split()

# NLTK tokenization without requiring punkt_tab
nltk_tokens = word_tokenize(clean_text, preserve_line=True)

print("Python split():")
print(split_tokens)

print("\nNLTK word_tokenize():")
print(nltk_tokens)

print("\nNumber of tokens using split():", len(split_tokens))
print("Number of tokens using word_tokenize():", len(nltk_tokens))
```

```python
stop_words = {
    'a', 'an', 'the', 'is', 'are', 'was', 'were', 'be', 'been',
    'being', 'and', 'or', 'but', 'if', 'to', 'of', 'in', 'on',
    'for', 'with', 'by', 'from', 'as', 'at', 'it', 'this', 'that',
    'these', 'those', 'can', 'will', 'their', 'they', 'them',
    'he', 'she', 'we', 'you', 'i', 'my', 'your', 'our', 'has',
    'have', 'had', 'do', 'does', 'did', 'through'
}

filtered_tokens = [
    word for word in word_tokens
    if word.isalpha() and word not in stop_words
]

print("Tokens After Removing Stopwords:")
print(filtered_tokens)
```

```python
from collections import Counter
import matplotlib.pyplot as plt

# Calculate word frequencies
frequency = Counter(filtered_tokens)

print("Word Frequency Distribution:\n")

for word, count in frequency.most_common():
    print(f"{word}: {count}")

# Plot the 10 most frequent words
top_words = frequency.most_common(10)

words = [item[0] for item in top_words]
counts = [item[1] for item in top_words]

plt.figure(figsize=(10, 5))
plt.bar(words, counts)
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("Top 10 Word Frequencies After Stopword Removal")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
texts = [
    "Artificial intelligence helps computers learn from data.",
    "Machine learning improves predictions using data.",
    "Artificial intelligence and machine learning power smart applications."
]

for i, text in enumerate(texts, start=1):
    print(f"Text {i}: {text}")
```

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

# Create the CountVectorizer
count_vectorizer = CountVectorizer()

# Convert texts into a Bag of Words matrix
bow_matrix = count_vectorizer.fit_transform(texts)

# Display the matrix as a DataFrame
bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=count_vectorizer.get_feature_names_out(),
    index=["Text 1", "Text 2", "Text 3"]
)

print("Bag of Words Representation:\n")
print(bow_df)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create the TF-IDF vectorizer
tfidf_vectorizer = TfidfVectorizer()

# Calculate TF-IDF scores
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

# Display the TF-IDF matrix
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out(),
    index=["Text 1", "Text 2", "Text 3"]
)

print("TF-IDF Representation:\n")
print(tfidf_df.round(3))
```

```python
feature_names = tfidf_vectorizer.get_feature_names_out()

for i in range(len(texts)):
    scores = tfidf_matrix[i].toarray().flatten()

    # Get indices of the three highest scores
    top_indices = scores.argsort()[::-1][:3]

    print(f"\nText {i + 1}: {texts[i]}")
    print("Top 3 Keywords:")

    for index in top_indices:
        print(f"{feature_names[index]}: {scores[index]:.3f}")
```

```python
text1 = """Artificial intelligence enables computers to perform intelligent tasks.
Machine learning helps systems learn from data.
AI supports automation, prediction, and decision making.
It is widely used in healthcare and education.
AI systems improve efficiency and solve complex problems."""

text2 = """Blockchain is a decentralized digital ledger.
It stores transactions in linked blocks.
Cryptography helps protect blockchain data.
Blockchain supports secure transactions without a central authority.
It is used in cryptocurrency and supply chain management."""

print("Text 1: Artificial Intelligence\n")
print(text1)

print("\nText 2: Blockchain\n")
print(text2)
```

```python
import re

def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    tokens = text.split()
    return tokens

tokens1 = preprocess(text1)
tokens2 = preprocess(text2)

print("AI Tokens:")
print(tokens1)

print("\nBlockchain Tokens:")
print(tokens2)
```

```python
# Convert token lists into sets
set1 = set(tokens1)
set2 = set(tokens2)

# Find common and combined unique words
intersection = set1.intersection(set2)
union = set1.union(set2)

# Calculate Jaccard similarity
jaccard_similarity = len(intersection) / len(union)

print("Common Words:", intersection)
print("Number of Common Words:", len(intersection))
print("Total Unique Words:", len(union))
print("Jaccard Similarity:", round(jaccard_similarity, 4))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Convert both texts into TF-IDF vectors
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform([text1, text2])

# Calculate cosine similarity
cosine_score = cosine_similarity(
    tfidf_matrix[0:1],
    tfidf_matrix[1:2]
)[0][0]

print("Cosine Similarity:", round(cosine_score, 4))
```

```python
print("Similarity Comparison")
print("---------------------")
print("Jaccard Similarity:", round(jaccard_similarity, 4))
print("Cosine Similarity:", round(cosine_score, 4))

if jaccard_similarity > cosine_score:
    print("Jaccard Similarity is higher.")
elif cosine_score > jaccard_similarity:
    print("Cosine Similarity is higher.")
else:
    print("Both similarity scores are equal.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import re

reviews = [
    "This phone is amazing and the battery life is excellent.",
    "The camera is fantastic and the performance is very smooth.",
    "I love this product because it is fast and reliable.",
    "The phone is terrible and the battery drains quickly.",
    "The product is disappointing and the screen quality is poor.",
    "The device is slow and the customer service is awful.",
    "The phone is okay and has an average battery.",
    "The product is neither good nor bad.",
    "The device is acceptable but nothing special."
]

df = pd.DataFrame({"Review": reviews})
```

```python
positive_words = {
    "amazing", "excellent", "fantastic", "smooth", "love",
    "fast", "reliable", "good", "great", "wonderful", "best"
}

negative_words = {
    "terrible", "drains", "quickly", "disappointing", "poor",
    "slow", "awful", "bad", "worst", "hate"
}

def analyze_sentiment(review):
    words = re.findall(r"\b[a-z]+\b", review.lower())

    positive_count = sum(word in positive_words for word in words)
    negative_count = sum(word in negative_words for word in words)

    if positive_count > negative_count:
        return "Positive"
    elif negative_count > positive_count:
        return "Negative"
    else:
        return "Neutral"

df["Sentiment"] = df["Review"].apply(analyze_sentiment)

print(df)
```

```python
sentiment_counts = df["Sentiment"].value_counts()

sentiment_counts.plot(
    kind="bar",
    title="Sentiment Distribution of Reviews",
    xlabel="Sentiment",
    ylabel="Number of Reviews"
)

plt.xticks(rotation=0)
plt.show()
```

```python
from collections import Counter

positive_reviews = df[df["Sentiment"] == "Positive"]["Review"]

positive_text = " ".join(positive_reviews).lower()
words = re.findall(r"\b[a-z]+\b", positive_text)

stopwords = {
    "the", "and", "is", "this", "it", "a", "i",
    "because", "has", "very", "life"
}

word_counts = Counter(
    word for word in words if word not in stopwords
)

if word_counts:
    common_words = word_counts.most_common(10)
    labels, counts = zip(*common_words)

    plt.figure(figsize=(10, 5))
    plt.bar(labels, counts)
    plt.title("Most Frequent Words in Positive Reviews")
    plt.xlabel("Words")
    plt.ylabel("Frequency")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
else:
    print("No positive-review words found.")
```

```python
import random
import re

text = """
Artificial intelligence helps computers learn from data.
Machine learning enables computers to make predictions.
Artificial intelligence is transforming modern technology.
Machine learning helps improve intelligent systems.
Computers learn patterns from data and make predictions.
"""
```

```python
words = re.findall(r"\b[\w']+\b", text.lower())

markov_chain = {}

for i in range(len(words) - 1):
    current_word = words[i]
    next_word = words[i + 1]

    if current_word not in markov_chain:
        markov_chain[current_word] = []

    markov_chain[current_word].append(next_word)

print(markov_chain)
```

```python
def generate_text(start_word, length=20):
    current_word = start_word.lower()
    generated_words = [current_word]

    for _ in range(length - 1):
        next_words = markov_chain.get(current_word)

        if not next_words:
            break

        current_word = random.choice(next_words)
        generated_words.append(current_word)

    return " ".join(generated_words)

print(generate_text("artificial", 20))
```
